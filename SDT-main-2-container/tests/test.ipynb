```python
import os

def get_bom_json_paths(directory):
    bom_paths = []
    for root, dirs, files in os.walk(directory):
        for name in files:
            if name == 'bom.json':
                bom_paths.append(os.path.join(root, name))
    return bom_paths

# Example usage
directory = '/home/abdul/DigitalTwin/Cobalt/SDT/SDT/test_files/bom-examples-master/'
bom_json_paths = get_bom_json_paths(directory)

# Save the paths to a file
output_file = 'bom_json_paths.txt'
with open(output_file, 'w') as file:
    for path in bom_json_paths:
        file.write(path + '\n')

print(f"Paths saved to {output_file}")
```

```python
with open(output_file, 'r') as file:
    first_path = file.readline().strip()

print(f"The first file path is: {first_path}")
```

```python
from json import dump as json_dump, load as json_loads, loads as json_loads

with open(first_path, 'r') as file:
    cyclonedx_json = file.read()

def convert_cyclonedx_to_wot(cyclonedx_json):
    cyclonedx_data = json_loads(cyclonedx_json)
    wot_data = {
        "id": cyclonedx_data.get("serialNumber", ""),
        "title": cyclonedx_data.get("metadata", {}).get("component", {}).get("name", ""),
        "description": cyclonedx_data.get("metadata", {}).get("component", {}).get("description", ""),
        "properties": {},
        "actions": {},
        "events": {}
    }

    for component in cyclonedx_data.get("components", []):
        component_id = component.get("bom-ref", "")
        wot_data["properties"][component_id] = {
            "type": component.get("type", ""),
            "name": component.get("name", ""),
            "description": component.get("description", ""),
            "properties": {prop["name"]: prop["value"] for prop in component.get("properties", [])}
        }

    return wot_data

# Example usage
wot_data = convert_cyclonedx_to_wot(cyclonedx_json)
print(wot_data)
# Save the WoT data to a file

wot_output_file = 'wot_data.json'
with open(wot_output_file, 'w') as file:
    json_dump(wot_data, file, indent=2)

print(f"WoT data saved to {wot_output_file}")
```
